# 05 — Hypnagogic precision tune

Wave 2 step 1: raise hypnagogic precision on SC4002 holdout via **threshold sweep on SC4001 val** (no SC4002 leakage for selection).

Preferred: `python scripts/tune_hypnagogic_threshold.py` then read metrics below.


```python
from pathlib import Path
import json
ROOT = Path('/workspace/muse-eeg-heads')
if not (ROOT / 'exports/head_a_holdout/precision_tune_metrics.json').exists():
    ROOT = Path('.').resolve()
print('ROOT', ROOT)
metrics = json.loads((ROOT / 'exports/head_a_holdout/precision_tune_metrics.json').read_text())
sel = metrics['selected']
imp = metrics['improvement']
print('method', sel['method'], 'threshold', sel['threshold'])
print('SC4002 hypnagogic precision', imp['sc4002_all_hypnagogic_precision'])
print('SC4002 hypnagogic recall', imp['sc4002_all_hypnagogic_recall'])
print('SC4002 macro-F1', imp['sc4002_all_macro_f1'])
```

## Before / after table


```python
rows = []
before_map = {
    'SC4002 full': metrics['before']['sc4002_all_labeled'],
    'SC4002 balanced': metrics['before']['sc4002_balanced'],
    'SC4001 val': metrics['before']['sc4001_val'],
}
after_map = {
    'SC4002 full': sel['sc4002_all_labeled'],
    'SC4002 balanced': sel['sc4002_balanced'],
    'SC4001 val': sel['sc4001_val'],
}
for label in before_map:
    for tag, m in [('before', before_map[label]), ('after', after_map[label])]:
        h = m['per_class']['hypnagogic']
        rows.append({
            'split': label, 'when': tag,
            'hypn_P': round(h['precision'], 3), 'hypn_R': round(h['recall'], 3),
            'hypn_F1': round(h['f1'], 3), 'macro_F1': round(m['macro_f1'], 3),
            'confusion': m['confusion'],
        })
try:
    import pandas as pd
    display(pd.DataFrame(rows))
except Exception:
    from pprint import pprint; pprint(rows)
```
